# English → Tamil Seq2Seq — modified from `nlp_chat(1).ipynb`

This version follows the **friend's working chatbot architecture**:

`Tokenizer → Keras Embedding → Encoder LSTM → Decoder LSTM → Dense`

**Word2Vec is removed completely.**

It is adapted for the 101-pair English→Tamil dataset and fixes the decoder teacher-forcing and inference flow.


```python
import numpy as np
import pandas as pd
import tensorflow as tf

from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, LSTM, Embedding, Dense
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

print("TensorFlow:", tf.__version__)
```

```python
# ============================================================
# 1. LOAD DATASET
# ============================================================

train = pd.read_csv("en_ta_small_clean_test_dataset.csv")

# Support both column names used in your files:
#   english / tamil
#   en / ta

if "english" in train.columns and "tamil" in train.columns:
    english_sentences = train["english"].astype(str).tolist()
    tamil_sentences = train["tamil"].astype(str).tolist()
elif "en" in train.columns and "ta" in train.columns:
    english_sentences = train["en"].astype(str).tolist()
    tamil_sentences = train["ta"].astype(str).tolist()
else:
    raise ValueError(
        f"Expected columns ['english','tamil'] or ['en','ta']. Found: {list(train.columns)}"
    )

# Remove empty rows and duplicates
pairs = pd.DataFrame({
    "english": english_sentences,
    "tamil": tamil_sentences
})

pairs = pairs.dropna()
pairs = pairs[
    (pairs["english"].str.strip() != "") &
    (pairs["tamil"].str.strip() != "")
]
pairs = pairs.drop_duplicates().reset_index(drop=True)

english_sentences = pairs["english"].tolist()
tamil_sentences = pairs["tamil"].tolist()

print("Total pairs:", len(pairs))
print()
print("Example:")
print("EN:", english_sentences[0])
print("TA:", tamil_sentences[0])
```

## 2. Add `<SOS>` and `<EOS>` only to the Tamil decoder side

This follows your friend's chatbot design.

The English encoder does **not** need `<SOS>` / `<EOS>` for this experiment.

Tamil target:

`<SOS> Tamil sentence <EOS>`


```python
# ============================================================
# 2. ADD SOS / EOS TO TAMIL TARGETS
# ============================================================

SOS = "<sos>"
EOS = "<eos>"

tamil_sent_SE = [
    f"{SOS} {sentence} {EOS}"
    for sentence in tamil_sentences
]

print("English:")
print(english_sentences[0])

print("\nTamil target:")
print(tamil_sent_SE[0])
```

```python
# ============================================================
# 3. TOKENIZERS
# ============================================================

english_tokenizer = Tokenizer(
    filters="",
    oov_token="<unk>"
)

tamil_tokenizer = Tokenizer(
    filters="",
    oov_token="<unk>"
)

english_tokenizer.fit_on_texts(english_sentences)
tamil_tokenizer.fit_on_texts(tamil_sent_SE)

english_sequences = english_tokenizer.texts_to_sequences(
    english_sentences
)

tamil_sequences = tamil_tokenizer.texts_to_sequences(
    tamil_sent_SE
)

english_vocab_size = len(english_tokenizer.word_index) + 1
tamil_vocab_size = len(tamil_tokenizer.word_index) + 1

print("English vocabulary:", english_vocab_size)
print("Tamil vocabulary:", tamil_vocab_size)
```

```python
# ============================================================
# 4. MAXIMUM SEQUENCE LENGTHS
# ============================================================

max_input_seq_length = max(
    len(x) for x in english_sequences
)

max_output_seq_length = max(
    len(x) for x in tamil_sequences
)

print("Max English length:", max_input_seq_length)
print("Max Tamil length:", max_output_seq_length)
```

```python
# ============================================================
# 5. PADDING
# ============================================================

input_sequences = pad_sequences(
    english_sequences,
    maxlen=max_input_seq_length,
    padding="post"
)

output_sequences = pad_sequences(
    tamil_sequences,
    maxlen=max_output_seq_length,
    padding="post"
)

print("Input shape:", input_sequences.shape)
print("Output shape:", output_sequences.shape)
```

## 6. Decoder input / target

For:

`<sos> செயல்படுத்துவதற்கான காலக்கெடு ஒரு வருடம் <eos>`

we train:

**Decoder input**

`<sos> செயல்படுத்துவதற்கான காலக்கெடு ஒரு வருடம்`

**Decoder target**

`செயல்படுத்துவதற்கான காலக்கெடு ஒரு வருடம் <eos>`

This is the important teacher-forcing shift.


```python
# ============================================================
# 6. DECODER INPUT / TARGET
# ============================================================

decoder_input_sequences = output_sequences[:, :-1]
decoder_target_ids = output_sequences[:, 1:]

print("Decoder input shape:", decoder_input_sequences.shape)
print("Decoder target shape:", decoder_target_ids.shape)
```

```python
# ============================================================
# 7. ONE-HOT TARGETS + PADDING MASK
# ============================================================

decoder_output_sequences = tf.keras.utils.to_categorical(
    decoder_target_ids,
    num_classes=tamil_vocab_size
)

# Padding token ID is 0.
# Do not allow padded positions to contribute to the loss/accuracy.
sample_weight = (decoder_target_ids != 0).astype("float32")

print("Decoder output shape:", decoder_output_sequences.shape)
print("Sample-weight shape:", sample_weight.shape)
```

```python
# ============================================================
# 8. VERIFY ONE EXAMPLE
# ============================================================

tamil_index_word = {
    index: word
    for word, index in tamil_tokenizer.word_index.items()
}

i = 0

print("English:")
print(english_sentences[i])

print("\nFull Tamil:")
print(tamil_sent_SE[i])

print("\nDecoder input:")
print([
    tamil_index_word.get(int(x), "<UNK>")
    for x in decoder_input_sequences[i]
    if x != 0
])

print("\nDecoder target:")
print([
    tamil_index_word.get(int(x), "<UNK>")
    for x in decoder_target_ids[i]
    if x != 0
])
```

## 9. Encoder

This is the same basic idea as your friend's notebook.

No Word2Vec:

`Token ID → Keras Embedding(64) → LSTM`


```python
# ============================================================
# 9. ENCODER
# ============================================================

EMBED_DIM = 64
LSTM_UNITS = 128

encoder_input = Input(
    shape=(max_input_seq_length,),
    name="encoder_input"
)

encoder_embedding = Embedding(
    input_dim=english_vocab_size,
    output_dim=EMBED_DIM,
    mask_zero=True,
    name="encoder_embedding"
)(encoder_input)

encoder_lstm = LSTM(
    LSTM_UNITS,
    return_state=True,
    name="encoder_lstm"
)

_, state_h, state_c = encoder_lstm(
    encoder_embedding
)
```

## 10. Decoder

The Tamil embedding is also learned by Keras.

There is **no Tamil Word2Vec model** here.


```python
# ============================================================
# 10. DECODER
# ============================================================

decoder_input_layer = Input(
    shape=(max_output_seq_length - 1,),
    name="decoder_input"
)

decoder_embedding_layer = Embedding(
    input_dim=tamil_vocab_size,
    output_dim=EMBED_DIM,
    mask_zero=True,
    name="decoder_embedding"
)

decoder_embedding = decoder_embedding_layer(
    decoder_input_layer
)

decoder_lstm = LSTM(
    LSTM_UNITS,
    return_sequences=True,
    return_state=True,
    name="decoder_lstm"
)

decoder_output, _, _ = decoder_lstm(
    decoder_embedding,
    initial_state=[state_h, state_c]
)

decoder_dense = Dense(
    tamil_vocab_size,
    activation="softmax",
    name="decoder_dense"
)

output = decoder_dense(decoder_output)
```

```python
# ============================================================
# 11. BUILD MODEL
# ============================================================

model = Model(
    [encoder_input, decoder_input_layer],
    output
)

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="categorical_crossentropy",
    weighted_metrics=["accuracy"]
)

model.summary()
```

## 12. Train

For this first experiment, the goal is **memorization/debugging**, not generalization.

We want to prove that this architecture can learn the 101 clean pairs.


```python
# ============================================================
# 12. TRAIN
# ============================================================

EPOCHS = 200
BATCH_SIZE = 16

history = model.fit(
    [input_sequences, decoder_input_sequences],
    decoder_output_sequences,
    sample_weight=sample_weight,
    batch_size=BATCH_SIZE,
    epochs=EPOCHS,
    shuffle=True,
    verbose=1
)
```

## 13. Build inference models

Training uses the complete decoder sequence.

Inference is different: start with `<SOS>`, predict one Tamil token, feed that token back, and continue until `<EOS>`.


```python
# ============================================================
# 13. INFERENCE MODELS
# ============================================================

encoder_model = Model(
    encoder_input,
    [state_h, state_c]
)

inf_decoder_input = Input(
    shape=(1,),
    name="inf_decoder_input"
)

inf_state_h = Input(
    shape=(LSTM_UNITS,),
    name="inf_state_h"
)

inf_state_c = Input(
    shape=(LSTM_UNITS,),
    name="inf_state_c"
)

inf_embedding = decoder_embedding_layer(
    inf_decoder_input
)

inf_output, inf_h, inf_c = decoder_lstm(
    inf_embedding,
    initial_state=[inf_state_h, inf_state_c]
)

inf_probs = decoder_dense(
    inf_output
)

decoder_model = Model(
    [inf_decoder_input, inf_state_h, inf_state_c],
    [inf_probs, inf_h, inf_c]
)

sos_id = tamil_tokenizer.word_index[SOS]
eos_id = tamil_tokenizer.word_index[EOS]

print("SOS ID:", sos_id)
print("EOS ID:", eos_id)
```

```python
# ============================================================
# 14. ENGLISH → TAMIL TRANSLATION
# ============================================================

def translate(sentence, max_len=None):

    if max_len is None:
        max_len = max_output_seq_length

    # English -> token IDs
    sequence = english_tokenizer.texts_to_sequences(
        [sentence]
    )

    sequence = pad_sequences(
        sequence,
        maxlen=max_input_seq_length,
        padding="post"
    )

    # Encoder states
    h, c = encoder_model.predict(
        sequence,
        verbose=0
    )

    # Start with SOS
    current_token = np.array(
        [[sos_id]],
        dtype=np.int32
    )

    result = []

    for _ in range(max_len):

        prediction, h, c = decoder_model.predict(
            [current_token, h, c],
            verbose=0
        )

        predicted_id = int(
            np.argmax(prediction[0, -1, :])
        )

        # Stop at EOS or padding
        if predicted_id == eos_id or predicted_id == 0:
            break

        predicted_word = tamil_index_word.get(
            predicted_id,
            "<unk>"
        )

        if predicted_word not in (SOS, EOS):
            result.append(predicted_word)

        # Feed prediction into next decoder step
        current_token = np.array(
            [[predicted_id]],
            dtype=np.int32
        )

    return " ".join(result)
```

## 15. Test a training example

First test a sentence that the model has already seen.

For example:

`The time frame for implementation is one year.`

Because this is a tiny debugging dataset, the first objective is for the model to reproduce its training examples.


```python
# ============================================================
# 15. TEST
# ============================================================

test_sentence = "The time frame for implementation is one year."

print("English:")
print(test_sentence)

print("\nTamil:")
print(translate(test_sentence))
```

```python
# ============================================================
# 16. TEST MORE TRAINING EXAMPLES
# ============================================================

for i in [0, 1, 2, 10, 20]:

    print("\n" + "=" * 60)
    print("English :", english_sentences[i])
    print("Expected:", tamil_sentences[i])
    print("Model   :", translate(english_sentences[i]))
```

## What success looks like

For the first test, we want something close to:

`The time frame for implementation is one year.`

→

`செயல்படுத்துவதற்கான காலக்கெடு ஒரு வருடம்`

If it still produces:

`இந்த மற்றும் மற்றும் மற்றும்...`

then the next thing to inspect is the actual training loss and token probabilities — not another dataset.

**Important:** 200 epochs on 101 examples is intentionally a debugging experiment. It is not a production translation model.
