```python
import pandas as pd
import numpy as np
```

```python
#Function to add SOS and EOS to the statement
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.layers import Embedding, LSTM, Dense
from tensorflow.keras.models import Sequential

train=pd.read_csv("en_ta_small_clean_test_dataset.csv")
# train=train.drop(["Unnamed: 0"],axis=1)
# english_sentences=train["en"]
# tamil_sentence=train['ta']
english_sentences=train["english"]
tamil_sentence=train['tamil']
english_sentences=english_sentences
tamil_sentences=tamil_sentence

def addSosEos(seriesSentence):
    # Define the <SOS> and <EOS> tokens
    sos_token = "<SOS>"
    eos_token = "<EOS>"

    # Add <SOS> and <EOS> tokens to each statement
    statements_with_tokens = [f"{sos_token} {statement} {eos_token}" for statement in seriesSentence]

    english_sent=[]
    # Print the statements with tokens
    for statement in statements_with_tokens:
        english_sent.append(statement)
        # print(statement)
    return english_sent
```

```python
english_sent_SE=addSosEos(english_sentences)
tamil_sent_SE=addSosEos(tamil_sentences)
```

```python
print(english_sent_SE[:5])
print(tamil_sent_SE[:5])
```

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, LSTM, Embedding, Dense
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
# Tokenize the English and Tamil sentences
english_tokenizer = Tokenizer(filters="")
english_tokenizer.fit_on_texts(english_sent_SE)
english_vocab_size = len(english_tokenizer.word_index) + 1
english_sequences = english_tokenizer.texts_to_sequences(english_sent_SE)
```

```python
english_vocab_size
```

```python
tamil_tokenizer = Tokenizer(filters="")
tamil_tokenizer.fit_on_texts(tamil_sent_SE)
tamil_vocab_size = len(tamil_tokenizer.word_index) + 1
tamil_sequences = tamil_tokenizer.texts_to_sequences(tamil_sent_SE)
```

```python
tamil_vocab_size
```

```python
max_input_seq_length=10
max_output_seq_length=10
```

```python
# Pad sequences to a fixed length
input_sequences = pad_sequences(english_sequences, maxlen=max_input_seq_length, padding='post')
output_sequences = pad_sequences(tamil_sequences, maxlen=max_output_seq_length, padding='post')
```

```python
# Prepare the decoder input and output sequences for teacher forcing
# decoder_input_sequences = np.zeros_like(output_sequences)
# decoder_input_sequences[:, 1:] = output_sequences[:, :-1]
# decoder_input_sequences[:, 0] = tamil_tokenizer.word_index['<sos>']
# decoder_output_sequences = np.eye(tamil_vocab_size)[output_sequences]

#New

# ============================================================
# Decoder Input / Decoder Target
# ============================================================

decoder_input_sequences = np.zeros_like(output_sequences)
decoder_output_sequences = np.zeros_like(output_sequences)

# Decoder input:
# <SOS> word1 word2 word3 ...
decoder_input_sequences[:, :-1] = output_sequences[:, :-1]

# Decoder target:
# word1 word2 word3 ... <EOS>
decoder_output_sequences[:, :-1] = output_sequences[:, 1:]

# Convert target IDs to one-hot
decoder_output_sequences = np.eye(tamil_vocab_size)[decoder_output_sequences]
```

```python
from gensim.models import Word2Vec

eng_model = Word2Vec.load('engmodel.bin')
tam_model = Word2Vec.load('tammodel.bin')
```

```python
def create_embedding_matrix(word2vec_model, tokenizer, vocab_size):
    embedding_matrix = np.zeros((vocab_size, word2vec_model.vector_size))
    for word, i in tokenizer.word_index.items():
        try:
            embedding_vector = word2vec_model.wv[word]
            embedding_matrix[i] = embedding_vector
        except KeyError:
            pass  # Words not found in the embedding index will be all zeros
    return embedding_matrix

eng_embedding_matrix = create_embedding_matrix(eng_model, english_tokenizer, english_vocab_size)
tam_embedding_matrix = create_embedding_matrix(tam_model, tamil_tokenizer, tamil_vocab_size)
```

```python
eng_embedding_matrix.shape
```

```python
tam_embedding_matrix.shape
```

```python
def create_seq2seq_model(input_vocab_size, output_vocab_size, input_seq_length, output_seq_length, hidden_units, eng_embedding_matrix, tam_embedding_matrix):
    # Encoder
    encoder_inputs = Input(shape=(input_seq_length,))
    encoder_embedding = Embedding(input_vocab_size, hidden_units, weights=[eng_embedding_matrix], trainable=True)(encoder_inputs)
    encoder_lstm, encoder_state_h, encoder_state_c = LSTM(hidden_units, return_state=True)(encoder_embedding)

    # Decoder
    decoder_inputs = Input(shape=(output_seq_length,))
    decoder_embedding = Embedding(output_vocab_size, hidden_units, weights=[tam_embedding_matrix], trainable=True)(decoder_inputs)
    decoder_lstm = LSTM(hidden_units, return_sequences=True, return_state=True)
    decoder_outputs, _, _ = decoder_lstm(decoder_embedding, initial_state=[encoder_state_h, encoder_state_c])
    decoder_dense = Dense(output_vocab_size, activation='softmax')
    decoder_outputs = decoder_dense(decoder_outputs)

    model = Model([encoder_inputs, decoder_inputs], decoder_outputs)
    return model
```

```python
# Convert target_sequences to one-hot encoded format
target_sequences = tf.keras.utils.to_categorical(output_sequences, num_classes=tamil_vocab_size)
```

```python
model = create_seq2seq_model(english_vocab_size, tamil_vocab_size, max_input_seq_length, max_output_seq_length, 128, eng_embedding_matrix, tam_embedding_matrix)
```

```python
# Compile the model
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
```

```python
import numpy as np
from tensorflow.keras.preprocessing.sequence import pad_sequences


# ============================================================
# 1. INPUT SENTENCE
# ============================================================

input_sentence = ("<sos> The time frame for implementation is one year, <eos>")


# ============================================================
# 2. ENGLISH → TOKEN IDs
# ============================================================

input_sequence = english_tokenizer.texts_to_sequences(
    [input_sentence]
)

input_sequence = pad_sequences(
    input_sequence,
    maxlen=max_input_seq_length,
    padding="post"
)


# ============================================================
# 3. TAMIL INDEX → WORD
# ============================================================

tamil_index_word = {
    index: word
    for word, index in tamil_tokenizer.word_index.items()
}


# ============================================================
# 4. GET SOS / EOS IDs
# ============================================================

sos_id = tamil_tokenizer.word_index["<sos>"]
eos_id = tamil_tokenizer.word_index["<eos>"]


# ============================================================
# 5. START DECODER WITH <SOS>
# ============================================================

decoder_sequence = np.zeros(
    (1, max_output_seq_length),
    dtype=np.int32
)

decoder_sequence[0, 0] = sos_id


# ============================================================
# 6. AUTOREGRESSIVE DECODING
# ============================================================

decoded_words = []

for step in range(1, max_output_seq_length):

    predictions = model.predict(
        [input_sequence, decoder_sequence],
        verbose=0
    )

    # Prediction for current position
    predicted_id = np.argmax(
        predictions[0, step - 1, :]
    )

    # Stop at EOS
    if predicted_id == eos_id:
        break

    # Ignore padding
    if predicted_id == 0:
        break

    predicted_word = tamil_index_word.get(
        predicted_id,
        "<unk>"
    )

    # Ignore SOS
    if predicted_word == "<sos>":
        continue

    decoded_words.append(predicted_word)

    # Feed predicted token back to decoder
    decoder_sequence[0, step] = predicted_id


# ============================================================
# 7. FINAL TRANSLATION
# ============================================================

decoded_statement = " ".join(decoded_words)


# ============================================================
# 8. DISPLAY
# ============================================================

print("English:")
print(input_sentence)

print("\nTamil:")
print(decoded_statement)
```
