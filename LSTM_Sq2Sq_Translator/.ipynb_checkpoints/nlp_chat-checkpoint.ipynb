```python
# # ============================================================
# # 1. DATA
# # ============================================================

# conversation_groups = [

#     (["hello", "hi", "hey", "hello there", "hi there"],
#      "hello nice to meet you"),

#     (["good morning", "morning", "hello good morning"],
#      "good morning how are you"),

#     (["good evening", "evening", "hello good evening"],
#      "good evening how are you"),

#     (["how are you", "how are you doing", "are you okay"],
#      "i am doing well thank you"),

#     (["what is your name", "tell me your name", "your name"],
#      "my name is minibot"),

#     (["who are you", "what are you", "are you a chatbot"],
#      "i am minibot a simple chatbot"),

#     (["what can you do", "how can you help me", "what do you do"],
#      "i can chat with you and answer simple questions"),

#     (["thank you", "thanks", "thanks a lot"],
#      "you are welcome"),

#     (["bye", "goodbye", "see you", "see you later"],
#      "goodbye see you later"),

#     (["help me", "i need help", "can you help me"],
#      "of course tell me what you need help with"),

#     (["who created you", "who made you", "who built you"],
#      "i was created as a python learning project"),

#     (["tell me a joke", "make me laugh", "give me a joke"],
#      "why did the computer get cold because it left its windows open"),

#     (["how old are you", "what is your age"],
#      "i do not have an age like a human"),

#     (["how can i learn python", "teach me python", "is python easy"],
#      "start with python basics and practice every day"),

#     (["what is ai", "what is artificial intelligence", "explain ai"],
#      "artificial intelligence helps computers perform intelligent tasks"),

#     (["i am sad", "i feel sad", "i am unhappy"],
#      "i am sorry you feel that way"),

#     # E-commerce conversations
#     (["show me your products", "what products do you sell",
#       "open the product catalog", "i want to see your products"],
#      "you can browse our catalog to see the available products"),

#     (["help me find a product", "i am looking for a product",
#       "search for a product", "where can i find an item"],
#      "tell me the product name or category you are looking for"),

#     (["recommend a product", "what should i buy",
#       "help me choose a product", "show me popular products"],
#      "tell me your preferences and budget so i can suggest a product"),

#     (["what is the price", "how much does this cost",
#       "tell me the product price", "is this product expensive"],
#      "the current price is displayed on the product page"),

#     (["is this product available", "is this item in stock",
#       "do you have this product", "can i buy this item"],
#      "please check the product page for current stock availability"),

#     (["add this to my cart", "put this in my cart",
#       "add this item to my basket", "i want to add this product"],
#      "select the product options and use the add to cart button"),

#     (["how do i place an order", "how can i buy this",
#       "help me complete my purchase", "i want to order this product"],
#      "add the product to your cart and continue to checkout"),

#     (["where is my order", "track my order",
#       "what is my order status", "has my order shipped"],
#      "please provide your order number so its status can be checked"),

#     (["when will my order arrive", "what is the delivery date",
#       "how long does delivery take", "when will i get my package"],
#      "the estimated delivery date is shown in your order details"),

#     (["how much is shipping", "is shipping free",
#       "what is the delivery charge", "are there shipping fees"],
#      "shipping charges depend on your location and order value"),

#     (["what payment methods do you accept", "how can i pay",
#       "can i pay by card", "show me payment options"],
#      "the available payment methods are displayed during checkout"),

#     (["is cash on delivery available", "can i pay on delivery",
#       "do you accept cash on delivery", "i want cash on delivery"],
#      "cash on delivery availability depends on your location"),

#     (["my payment failed", "payment is not working",
#       "my payment was declined", "i cannot complete payment"],
#      "check your payment details or try another payment method"),

#     (["cancel my order", "i want to cancel my order",
#       "how can i cancel an order", "can my order be cancelled"],
#      "please provide your order number so cancellation can be checked"),

#     (["return my product", "i want to return my order",
#       "how can i return an item", "can i send this product back"],
#      "open your order details and select the return option"),

#     (["where is my refund", "when will i receive my refund",
#       "check my refund status", "how long does a refund take"],
#      "refund time depends on the payment method and your bank"),

#     (["my product arrived damaged", "the item is broken",
#       "i received a damaged package", "the product does not work"],
#      "please report the damage with your order number and product photos"),

#     (["i received the wrong product", "this is not what i ordered",
#       "you sent the wrong item", "my order has another product"],
#      "please report the incorrect item with your order number"),

#     (["do you have discounts", "are there special offers",
#       "show me current deals", "can i get a discount"],
#      "current discounts are displayed on the offers page"),

#     (["connect me to customer support", "i want to talk to a person",
#       "let me speak to an agent", "i need human assistance"],
#      "i will connect you with a customer support agent")
# ]


import json
import pandas as pd

file_names = [
    "minibot_individual_5000_conversations.json",
    "minibot_simple_normal_conversation.json",
    "minibot_natural_conversation.json"
]

all_conversations = []

for file_name in file_names:
    with open(file_name, "r", encoding="utf-8") as file:
        data = json.load(file)
        all_conversations.extend(data["conversations"])

dataset = pd.DataFrame(all_conversations)

dataset = dataset.drop_duplicates(
    subset=["question"]
).reset_index(drop=True)

questions = dataset["question"].tolist()
answers = dataset["answer"].tolist()

dataset.to_csv(
    "chat_data.csv",
    index=False,
    encoding="utf-8"
)

print("CSV created successfully")
print("Total conversations:", len(dataset))
```

```python
# ============================================================
# 2. CREATE MORE TRAINING EXAMPLES
# ============================================================

prefixes = [
    "",
    "please ",
    "hey minibot ",
    "can you tell me ",
    "i want to know "
]

suffixes = [
    "",
    " please",
    " for me",
    " right now"
]

questions = []
answers = []

for row in dataset.itertuples(index=False):
    question = str(row.question).strip()
    answer = str(row.answer).strip()

    for prefix in prefixes:
        for suffix in suffixes:
            new_question = f"{prefix}{question}{suffix}".strip()

            questions.append(new_question)
            answers.append(answer)

print("Total examples:", len(questions))

print("\nExample:")
print(questions[0], "->", answers[0])
```

```python
# ============================================================
# 3. ADD START / END
# ============================================================

answers = [
    "<start> " + answer + " <end>"
    for answer in answers
]
```

```python
# ============================================================
# 4. TOKENIZER
# ============================================================

question_tokenizer = Tokenizer(
    filters="",
    oov_token="<unk>"
)

answer_tokenizer = Tokenizer(
    filters="",
    oov_token="<unk>"
)


question_tokenizer.fit_on_texts(questions)

answer_tokenizer.fit_on_texts(answers)
```

```python
# ============================================================
# 5. TEXT -> NUMBERS
# ============================================================

question_sequences = question_tokenizer.texts_to_sequences(
    questions
)

answer_sequences = answer_tokenizer.texts_to_sequences(
    answers
)
```

```python
# ============================================================
# 6. MAX LENGTH
# ============================================================

max_question_length = max(
    len(x) for x in question_sequences
)

max_answer_length = max(
    len(x) for x in answer_sequences
)

print(max_question_length)
```

```python
# ============================================================
# 7. PADDING
# ============================================================

question_sequences = pad_sequences(
    question_sequences,
    maxlen=max_question_length,
    padding="post"
)

answer_sequences = pad_sequences(
    answer_sequences,
    maxlen=max_answer_length,
    padding="post"
)
```

```python
# ============================================================
# 8. DECODER INPUT / TARGET
# ============================================================

decoder_input = answer_sequences[:, :-1]

decoder_target = answer_sequences[:, 1:]
```

```python
# ============================================================
# 9. VOCABULARY SIZE
# ============================================================

question_vocab = len(
    question_tokenizer.word_index
) + 1

answer_vocab = len(
    answer_tokenizer.word_index
) + 1


print("Question vocabulary:", question_vocab)
print("Answer vocabulary:", answer_vocab)
```

```python
# ============================================================
# 10. ENCODER
# ============================================================

encoder_input = Input(
    shape=(max_question_length,)
)


encoder_embedding = Embedding(
    question_vocab,
    64,
    mask_zero=True
)(encoder_input)


_, state_h, state_c = LSTM(
    128,
    return_state=True
)(encoder_embedding)
```

```python
# ============================================================
# 11. DECODER
# ============================================================

decoder_input_layer = Input(
    shape=(max_answer_length - 1,)
)


decoder_embedding = Embedding(
    answer_vocab,
    64,
    mask_zero=True
)(decoder_input_layer)


decoder_output = LSTM(
    128,
    return_sequences=True
)(
    decoder_embedding,
    initial_state=[state_h, state_c]
)
```

```python
# ============================================================
# 12. OUTPUT
# ============================================================

output = Dense(
    answer_vocab,
    activation="softmax"
)(decoder_output)
```

```python
# ============================================================
# 13. MODEL
# ============================================================

model = Model(
    [encoder_input, decoder_input_layer],
    output
)


model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)


model.summary()
```

```python
# ============================================================
# 14. TRAIN
# ============================================================

model.fit(
    [question_sequences, decoder_input],
    decoder_target,
    epochs=80,
    batch_size=32,
    validation_split=0.2
)
```

```python
# ============================================================
# 15. NUMBER -> WORD
# ============================================================

index_to_word = {
    index: word
    for word, index
    in answer_tokenizer.word_index.items()
}
```

```python

supported_questions = set(questions)

def chat(text):

    # Find similar question
    match = get_close_matches(
        text.lower().strip(),
        supported_questions,
        n=1,
        cutoff=0.65
    )


    if not match:
        return "I do not understand that yet"

    # Convert question to numbers
    question = question_tokenizer.texts_to_sequences([match[0]])

    question = pad_sequences(
        question,
        maxlen=max_question_length,
        padding="post"
    )



    # Start answer
    generated = [
        answer_tokenizer.word_index["<start>"]
    ]

    result = []

    # Generate answer word by word
    for i in range(max_answer_length - 1):

        decoder = pad_sequences(
            [generated],
            maxlen=max_answer_length - 1,
            padding="post"
        )

        prediction = model.predict(
            [question, decoder],
            verbose=0
        )

        word_id = np.argmax(prediction[0, i])

        word = index_to_word.get(word_id, "")

        if word == "<end>" or word == "":
            break

        result.append(word)
        generated.append(word_id)

    return " ".join(result)
```

```python
# ============================================================
# 17. START CHAT
# ============================================================

print("\n========================")
print("CHATBOT STARTED")
print("========================")


while True:

    user = input("\nYou: ")


    if user.lower() == "exit":

        print("Bot: Goodbye!")

        break


    response = chat(user)


    print("Bot:", response)
    
```
